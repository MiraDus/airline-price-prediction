```python
import requests
import pandas as pd

API_TOKEN = "YOUR_TRAVELPAYOUTS_TOKEN"
URL = "http://api.travelpayouts.com/v1/prices/calendar"

routes = [
    ("ALA", "NQZ"),  
    ("ALA", "SCO"),  
    ("ALA", "CIT"),  
    ("ALA", "URA"),  
    ("ALA", "UKK"),  
    ("ALA", "KZO"),  
    ("ALA", "PLX"),  
    ("ALA", "PWQ"),  
    ("ALA", "DMB"),  
    ("ALA", "EKB"),  

    ("NQZ", "ALA"),  
    ("NQZ", "SCO"),  
    ("NQZ", "CIT"),  
    ("NQZ", "URA"),  
    ("NQZ", "UKK"),  
    ("NQZ", "KZO"),  
    ("NQZ", "PLX"),  
    ("NQZ", "PWQ"),  
    ("NQZ", "DMB"),  
    ("NQZ", "EKB"),  

    ("SCO", "ALA"),  
    ("SCO", "NQZ"),  
    ("SCO", "CIT"),  
    ("SCO", "URA"),  
    ("SCO", "UKK"),  
    ("SCO", "KZO"),  
    ("SCO", "PLX"),  
    ("SCO", "PWQ"),  
    ("SCO", "DMB"),  
    ("SCO", "EKB"),  

    ("CIT", "ALA"),  
    ("CIT", "NQZ"),  
    ("CIT", "SCO"),  
    ("CIT", "URA"),  
    ("CIT", "UKK"),  
    ("CIT", "KZO"),  
    ("CIT", "PLX"),  
    ("CIT", "PWQ"),  
    ("CIT", "DMB"),  
    ("CIT", "EKB"),  

    ("URA", "ALA"),  
    ("URA", "NQZ"),  
    ("URA", "SCO"),  
    ("URA", "CIT"),  
    ("URA", "UKK"),  
    ("URA", "KZO"),  
    ("URA", "PLX"),  
    ("URA", "PWQ"),  
    ("URA", "DMB"),  
    ("URA", "EKB"),  

    ("UKK", "ALA"),  
    ("UKK", "NQZ"),  
    ("UKK", "SCO"),  
    ("UKK", "CIT"),  
    ("UKK", "URA"),  
    ("UKK", "KZO"),  
    ("UKK", "PLX"),  
    ("UKK", "PWQ"),  
    ("UKK", "DMB"),  
    ("UKK", "EKB"),  

    ("KZO", "ALA"),  
    ("KZO", "NQZ"),  
    ("KZO", "SCO"),  
    ("KZO", "CIT"),  
    ("KZO", "URA"),  
    ("KZO", "UKK"),  
    ("KZO", "PLX"),  
    ("KZO", "PWQ"),  
    ("KZO", "DMB"),  
    ("KZO", "EKB"),  

    ("PLX", "ALA"),  
    ("PLX", "NQZ"),  
    ("PLX", "SCO"),  
    ("PLX", "CIT"),  
    ("PLX", "URA"),  
    ("PLX", "UKK"),  
    ("PLX", "KZO"),  
    ("PLX", "PWQ"),  
    ("PLX", "DMB"),  
    ("PLX", "EKB"),  

    ("PWQ", "ALA"),  
    ("PWQ", "NQZ"),  
    ("PWQ", "SCO"),  
    ("PWQ", "CIT"),  
    ("PWQ", "URA"),  
    ("PWQ", "UKK"),  
    ("PWQ", "KZO"),  
    ("PWQ", "PLX"),  
    ("PWQ", "DMB"),  
    ("PWQ", "EKB"),  

    ("DMB", "ALA"),  
    ("DMB", "NQZ"),  
    ("DMB", "SCO"),  
    ("DMB", "CIT"),  
    ("DMB", "URA"),  
    ("DMB", "UKK"),  
    ("DMB", "KZO"),  
    ("DMB", "PLX"),  
    ("DMB", "PWQ"),  
    ("DMB", "EKB"),  

    ("EKB", "ALA"),  
    ("EKB", "NQZ"),  
    ("EKB", "SCO"),  
    ("EKB", "CIT"),  
    ("EKB", "URA"),  
    ("EKB", "UKK"),  
    ("EKB", "KZO"),  
    ("EKB", "PLX"),  
    ("EKB", "PWQ"),  
    ("EKB", "DMB"),  
    ("AKX", "ALA"),
("AKX", "NQZ"),
("AKX", "SCO"),
("AKX", "CIT"),
("AKX", "URA"),
("AKX", "UKK"),
("AKX", "KZO"),
("AKX", "PLX"),
("AKX", "PWQ"),
("AKX", "DMB"),
("AKX", "EKB"),
("AKX", "KSN"),
("AKX", "GUW"),
("AKX", "KOV"),
("AKX", "BXH"),
("AKX", "AYK"),
("AKX", "TDK"),
("AKX", "UZR"),
("AKX", "HSA"),
("AKX", "BPL"),

("TDK", "ALA"),
("TDK", "NQZ"),
("TDK", "SCO"),
("TDK", "CIT"),
("TDK", "URA"),
("TDK", "UKK"),
("TDK", "KZO"),
("TDK", "PLX"),
("TDK", "PWQ"),
("TDK", "DMB"),
("TDK", "EKB"),
("TDK", "KSN"),
("TDK", "GUW"),
("TDK", "KOV"),
("TDK", "BXH"),
("TDK", "AYK"),
("TDK", "UZR"),
("TDK", "HSA"),
("TDK", "BPL"),

("UZR", "ALA"),
("UZR", "NQZ"),
("UZR", "SCO"),
("UZR", "CIT"),
("UZR", "URA"),
("UZR", "UKK"),
("UZR", "KZO"),
("UZR", "PLX"),
("UZR", "PWQ"),
("UZR", "DMB"),
("UZR", "EKB"),
("UZR", "KSN"),
("UZR", "GUW"),
("UZR", "KOV"),
("UZR", "BXH"),
("UZR", "AYK"),
("UZR", "TDK"),
("UZR", "HSA"),
("UZR", "BPL"),

]

all_tickets = []

for origin, destination in routes:
    print(f"Запрашиваем данные: {origin} → {destination}")

    params = {
        "currency": "usd",
        "origin": origin,
        "destination": destination,
        "calendar_type": "departure_date",
        "token": API_TOKEN
    }

    response = requests.get(URL, params=params)

    if response.status_code == 200:
        data = response.json()

        if "data" in data and data["data"]:
            for date, flight in data["data"].items():
                all_tickets.append([
                    "Эконом",
                    flight.get("departure_at", date), 
                    origin,
                    destination,
                    flight.get("price", "Не указано"),
                    flight.get("airline", "Не указано"),
                    flight.get("flight_number", "Не указано"),
                    flight.get("return_at", "Не указано"),
                    flight.get("transfers", 0)
                ])

        print(f"✅ Данные получены: {len(all_tickets)} билетов!")

    else:
        print(f"❌ Ошибка {response.status_code}: {response.text}")

# ✅ Сохранение в CSV
df = pd.DataFrame(all_tickets, columns=[
    "Класс обслуживания", "Дата вылета", "Город отправления",
    "Город назначения", "Цена билета (USD)", "Авиакомпания",
    "Номер рейса", "Дата возвращения", "Количество пересадок"
])

df.to_csv("aviasales_calendar_flights.csv", index=False)
print(f"✅ Данные сохранены в aviasales_внутренние.csv! Всего билетов: {len(df)}")
```
